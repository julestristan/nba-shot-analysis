## 0. Charger la table et re-vérifier l'EDA

### 0.1 Imports et chemins

Le notebook est dans `notebooks/`, les données dans `data/` (ignoré par Git). Les 22 zips de saison sont lus directement, sans décompression sur le disque.

```python
import zipfile                      # lecture des CSV directement dans les zips
from pathlib import Path

import numpy as np
import pandas as pd

DATA   = Path("..") / "data"                         # dossier des données brutes (ignoré par Git)
SHOTS  = DATA / "shots_04_25"                        # les 22 zips de saison du dépôt DomSamangy/NBA_Shots_04_25
OUT    = DATA / "processed"                          # dossier de sortie du pre-processing
OUT.mkdir(parents=True, exist_ok=True)

fichiers = sorted(SHOTS.glob("NBA_*_Shots.csv.zip"))                     # un zip par saison, triés par année
print(len(fichiers), "fichiers de saison trouvés, de", fichiers[0].name, "à", fichiers[-1].name)
```

### 0.2 Charger les CSV depuis les archives

Un DataFrame par saison, puis concaténation en une seule table. Chaque zip contient un fichier parasite `__MACOSX/…` : seul le membre `.csv` est lu.

```python
def lire_saison(chemin_zip):
    """Lit le CSV contenu dans un zip de saison, en ignorant le fichier parasite __MACOSX."""
    with zipfile.ZipFile(chemin_zip) as archive:
        membres = [n for n in archive.namelist() if n.endswith(".csv") and not n.startswith("__MACOSX")]
        with archive.open(membres[0]) as f:                      # le seul vrai CSV de l'archive
            return pd.read_csv(f, low_memory=False)

morceaux = []                                                    # un DataFrame par saison
for f in fichiers:
    saison = lire_saison(f)
    morceaux.append(saison)
    print(f"{f.name:28s} {saison.shape[0]:>8,} lignes × {saison.shape[1]} colonnes".replace(",", " "))

df = pd.concat(morceaux, ignore_index=True)                      # empile les 22 tables en une table, aligne les colonnes par nom
del morceaux
```

`pd.concat` aligne les colonnes **par nom**, pas par position. Le point compte parce que le fichier **2024-25 n'a que 24 colonnes** (`POSITION_GROUP` et `POSITION` ne sont pas renseignées à la source pour cette saison). Alignées par nom, ces deux colonnes deviennent *manquantes* pour toute la saison 2024-25 : c'est l'origine exacte des 219 527 postes manquants constatés dans l'EDA. Une concaténation par position aurait décalé toutes les colonnes de cette saison et produit une table fausse sans message d'erreur.

`low_memory=False` fait lire chaque fichier en une fois, pour que le type de chaque colonne soit déterminé sans erreur. Durée : 20 à 60 secondes pour les 22 fichiers.

```python
# Contrôle rapide de la table assemblée : dimensions, colonnes, valeurs de la cible
print(df.shape)
print(list(df.columns))
print(df["SHOT_MADE"].unique()[:10])
```

### 0.3 Re-vérifier les chiffres de l'EDA

Avant toute transformation, cinq contrôles confirment que la table chargée est **la même** que celle de l'EDA (`output/eda.md`) :

1. 4 450 789 lignes × 26 colonnes ;
2. `SHOT_MADE` ne contient que `True` / `False` (des libellés comme `2PT Field Goal` dans cette colonne signaleraient un décalage de colonnes à la concaténation) ;
3. taux de réussite global 45,8 % ;
4. 227 457 postes manquants (219 527 pour toute la saison 2024-25 + 7 930 littéraux « NA » des saisons antérieures, lus comme manquants par pandas) ;
5. aucune valeur manquante ailleurs.

Ces contrôles ont rejeté une version « tout-en-un » du dataset diffusée sur le Drive de la source (24 colonnes, 4 443 714 lignes), qui n'est pas la table de l'EDA.

```python
def controle(nom, condition):
    """Affiche ok ou ko devant le nom du contrôle."""
    print(("ok" if condition else "ko"), nom)

na = df.isna().sum()                                         # nombre de valeurs manquantes par colonne

controle("4 450 789 lignes × 26 colonnes",          df.shape == (4_450_789, 26))
controle("SHOT_MADE ne contient que True/False",     set(df["SHOT_MADE"].unique()) <= {True, False})
controle("taux de réussite global = 45,8 %",         round(df["SHOT_MADE"].mean() * 100, 1) == 45.8)
controle("227 457 postes manquants",                 na["POSITION_GROUP"] == 227_457 and na["POSITION"] == 227_457)
controle("aucun autre NA",                           na.drop(["POSITION_GROUP", "POSITION"]).sum() == 0)

print()
print(df.dtypes.to_string())
```

**Lecture.** Les cinq contrôles passent. Côté types : `SHOT_MADE` est déjà booléen, `GAME_DATE` est du texte à convertir (section 3), et `POSITION` / `POSITION_GROUP` sont les seules colonnes avec des manquants.

## 1. Périmètre

### 1.1 Saison de début et fenêtre ≥ 2015-16

`SEASON_1` est l'**année de fin** de saison (2016 = saison 2015-16). On crée `saison_debut = SEASON_1 − 1`, convention usuelle (« saison 2015 » = celle qui commence à l'automne 2015).

**Pourquoi 2015-16** (cadrage) : le jeu NBA a basculé vers le tir à 3 points autour de 2013-2015 (EDA Kaggle G3 : pente de la part de 3 pts +0,54 pt/an jusqu'en 2012, +2,13 pt/an ensuite). Entraîner sur les années 2000 reviendrait à apprendre un jeu qui n'existe plus. Dix saisons restent, soit plus de 2 millions de tirs.

Effet induit : la **rupture de collecte des coordonnées** (EDA #1, saisons ≤ 2009-10) sort du périmètre.

```python
df["saison_debut"] = df["SEASON_1"] - 1                       # 2016 → 2015 (saison 2015-16)

n_avant = len(df)
d = df[df["saison_debut"] >= 2015].copy()                     # copie indépendante de df
del df                                                        # libère la mémoire de la table complète (≈ 2,5 Go)

print(f"Avant : {n_avant:,}  →  Après filtre ≥ 2015-16 : {len(d):,}  (retirées : {n_avant - len(d):,})".replace(",", " "))
print()
print("Tirs par saison dans la fenêtre :")
print(d["saison_debut"].value_counts().sort_index().to_string())
```

**Lecture.** 2 100 256 tirs sur 10 saisons, entre 188 116 (2019-20, saison COVID écourtée) et 219 527 (2024-25). Les volumes sont hétérogènes (EDA #5) : les comparaisons entre saisons se font en **taux**, jamais en volumes bruts.

### 1.2 Doublons exacts

Un doublon exact est une ligne identique sur les 26 colonnes. L'EDA en comptait 398 sur l'historique complet (EDA #6), attribués à des doubles saisies de la table de marque. On les affiche avant de les supprimer.

```python
masque_dup = d.duplicated(keep=False)                          # True sur TOUTES les lignes impliquées dans un doublon
print("Lignes impliquées dans un doublon :", masque_dup.sum(), "→ doublons à retirer :", d.duplicated().sum())
print()
colonnes_lecture = ["saison_debut", "GAME_ID", "PLAYER_NAME", "QUARTER", "MINS_LEFT", "SECS_LEFT",
                    "ACTION_TYPE", "SHOT_MADE", "LOC_X", "LOC_Y"]
print(d.loc[masque_dup, colonnes_lecture]
       .sort_values(["GAME_ID", "QUARTER", "MINS_LEFT", "SECS_LEFT"])
       .head(6).to_string(index=False))

n_avant = len(d)
d = d.drop_duplicates()                                        # garde la première occurrence de chaque ligne
print()
print(f"Avant : {n_avant:,}  →  Après suppression des doublons : {len(d):,}  (retirées : {n_avant - len(d):,})".replace(",", " "))
```

**Lecture.** 98 doublons dans la fenêtre (les 300 autres sont dans les saisons exclues). Exemple : *Tony Allen, Putback Layup, 11:30 au 2e quart-temps, raté, mêmes coordonnées*, deux fois. Deux putbacks ratés à la même seconde au même endroit ne sont pas deux tirs : c'est une double saisie. Suppression justifiée.

### 1.3 Tirs « Backcourt » et lignes « No Shot »

Deux catégories de lignes ne sont pas des tirs comparables aux autres :

- `BASIC_ZONE == "Backcourt"` : tirs déclenchés depuis sa propre moitié de terrain, 42 à 88 pieds du panier. Ce sont des tentatives désespérées de fin de quart-temps, pas des décisions de tir ; les garder fausserait l'extrémité de la relation distance-réussite ;
- `ACTION_TYPE == "No Shot"` : 81 lignes qui ne correspondent à aucun geste, probablement des événements mal codés.

Les effectifs sont mesurés avant le retrait.

```python
backcourt = d["BASIC_ZONE"] == "Backcourt"
print(f"Backcourt : {backcourt.sum():,} tirs ({backcourt.mean()*100:.2f} %) — "
      f"distance min {d.loc[backcourt, 'SHOT_DISTANCE'].min()} ft, médiane {d.loc[backcourt, 'SHOT_DISTANCE'].median():.0f} ft, "
      f"max {d.loc[backcourt, 'SHOT_DISTANCE'].max()} ft — réussite {d.loc[backcourt, 'SHOT_MADE'].mean()*100:.1f} %".replace(",", " "))

no_shot = d["ACTION_TYPE"] == "No Shot"
print(f"No Shot   : {no_shot.sum()} lignes — réussite {d.loc[no_shot, 'SHOT_MADE'].mean()*100:.1f} %")

n_avant = len(d)
d = d[~backcourt & ~no_shot].copy()                            # conserve les lignes ni Backcourt ni No Shot
print()
print(f"Avant : {n_avant:,}  →  Après retrait Backcourt + No Shot : {len(d):,}  (retirées : {n_avant - len(d):,})".replace(",", " "))

# Contrôle : lignes restantes étiquetées « Back Court Shot » par ZONE_RANGE
reste = d["ZONE_RANGE"] == "Back Court Shot"
print()
print(f"Lignes restantes avec ZONE_RANGE = 'Back Court Shot' : {reste.sum()} — zones : {d.loc[reste, 'BASIC_ZONE'].unique().tolist()}, "
      f"distance {d.loc[reste, 'SHOT_DISTANCE'].min()}–{d.loc[reste, 'SHOT_DISTANCE'].max()} ft, réussite {d.loc[reste, 'SHOT_MADE'].mean()*100:.1f} %")
```

**Lecture.** 4 439 tirs Backcourt (0,21 %, 2,3 % de réussite) et 81 « No Shot » retirés → **2 095 638 tirs**.

Le dernier contrôle montre que `ZONE_RANGE` et `BASIC_ZONE` ne sont pas parfaitement alignées : 453 tirs sont étiquetés « Back Court Shot » par la tranche de distance mais « Above the Break 3 » par la zone, à 39-47 ft (7,3 % de réussite). Ce sont des 3 points très lointains tentés depuis le camp offensif. Ils sont **conservés** : la zone fait foi, la tranche n'est qu'une discrétisation de la distance, retirée en section 5. L'effet sur le modèle est négligeable (0,02 % des lignes).

## 2. Fuites et redondances

Deux familles de colonnes à écarter avant toute modélisation :

- **Fuite** (*data leakage*) : une colonne connue *seulement après* le tir. Un modèle qui la voit « prédit » parfaitement en lisant la réponse. `EVENT_TYPE` (« Made Shot » / « Missed Shot ») est `SHOT_MADE` écrit en toutes lettres (EDA #4 : 0 incohérence sur 4,45 M) ;
- **Redondance** : deux colonnes qui portent la même information. Elles n'apportent pas de précision et perturbent certains modèles (régression logistique notamment). `SEASON_2` = `SEASON_1` en texte ; `ZONE_ABB` = `ZONE_NAME` abrégé ; `POSITION` (18 modalités dont des combinaisons `SG-PG`) se résume dans `POSITION_GROUP` (3 modalités).

Les équivalences sont re-vérifiées sur la fenêtre avant suppression.

```python
# 1) EVENT_TYPE doit être strictement équivalent à SHOT_MADE
incoherences = ((d["EVENT_TYPE"] == "Made Shot") != d["SHOT_MADE"]).sum()
controle(f"EVENT_TYPE <=> SHOT_MADE : {incoherences} incohérence(s)", incoherences == 0)

# 2) ZONE_NAME ⇔ ZONE_ABB : autant de paires distinctes que de modalités → correspondance 1 pour 1
paires = d.groupby(["ZONE_NAME", "ZONE_ABB"]).size().shape[0]
controle(f"ZONE_NAME ⇔ ZONE_ABB : {paires} paires pour {d['ZONE_NAME'].nunique()} modalités", paires == d["ZONE_NAME"].nunique())

# 3) SEASON_1 ⇔ SEASON_2
paires = d.groupby(["SEASON_1", "SEASON_2"]).size().shape[0]
controle(f"SEASON_1 ⇔ SEASON_2 : {paires} paires pour {d['SEASON_1'].nunique()} saisons", paires == d["SEASON_1"].nunique())

# 4) SHOT_TYPE (2 pts / 3 pts) vs BASIC_ZONE : cohérents ?
print()
print(pd.crosstab(d["BASIC_ZONE"], d["SHOT_TYPE"]))
```

**Lecture.** Les trois équivalences tiennent. Le tableau croisé `BASIC_ZONE × SHOT_TYPE` montre une poignée d'écarts (100 tirs à 2 pts dans « Above the Break 3 », 745 tirs à 3 pts dans « Mid-Range ») : un pied sur la ligne, zone attribuée à la position du pied et non du ballon. Négligeable (< 0,1 %). `SHOT_TYPE` est conservé comme **vérité du barème** (l'arbitre a compté 2 ou 3 points) et `BASIC_ZONE` comme **description de l'emplacement**.

```python
a_supprimer = ["EVENT_TYPE",   # fuite : c'est la cible écrite en toutes lettres
               "SEASON_2",     # redondant avec SEASON_1 / saison_debut
               "ZONE_ABB",     # redondant avec ZONE_NAME
               "POSITION"]     # trop fin (18 modalités), POSITION_GROUP suffit (EDA #2)
d = d.drop(columns=a_supprimer)
print("Colonnes restantes :", d.shape[1])
print(list(d.columns))
```

## 3. Conversions de type

- `SHOT_MADE` : booléen → entier **0 / 1**. Lève toute ambiguïté dans les calculs, et `mean()` donne directement le taux de réussite ;
- `GAME_DATE` : texte `MM-DD-YYYY` → date (EDA #9). Nécessaire pour trier chronologiquement et pour l'imputation des postes en 4.5 (« dernier poste connu ») ;
- `tir_3pts` : `SHOT_TYPE` → 0 / 1. Une catégorie à deux valeurs devient un indicateur.

```python
d["SHOT_MADE"] = d["SHOT_MADE"].astype(int)                                  # True → 1, False → 0
d["GAME_DATE"] = pd.to_datetime(d["GAME_DATE"], format="%m-%d-%Y")          # "11-12-2024" → 2024-11-12
d["tir_3pts"]  = (d["SHOT_TYPE"] == "3PT Field Goal").astype(int)            # 1 si tir à 3 points

print("Taux de réussite (fenêtre) :", round(d["SHOT_MADE"].mean() * 100, 1), "%")
print("Part de tirs à 3 points    :", round(d["tir_3pts"].mean() * 100, 1), "%")
print()
print("Première et dernière date de chaque saison :")
print(d.groupby("saison_debut")["GAME_DATE"].agg(["min", "max"]).to_string())
```

**Lecture.** Chaque saison va d'octobre à avril, sauf **2019-20**, terminée le 14 août 2020 (reprise dans la « bulle » d'Orlando après l'interruption COVID), et **2020-21**, commencée le 22 décembre. Les dates confirment le périmètre *saison régulière uniquement* : aucune date de mai-juin hors 2020-21, donc pas de playoffs.

## 4. Variables dérivées (*feature engineering*)

Principe : seules des variables **connues au moment du tir** sont créées (disponibilité a priori = Oui dans le template). Chaque variable répond à un constat de l'EDA.

### 4.1 Temps restant

`MINS_LEFT` et `SECS_LEFT` sont deux colonnes pour une seule information : le temps restant dans le quart-temps. Elles sont fusionnées en **secondes** (`temps_restant_qt`), puis complétées par le temps restant dans le **match** (`temps_restant_match`) : un tir à 2 minutes de la fin du 1er quart-temps et un tir à 2 minutes de la fin du 4e n'ont pas le même enjeu (EDA var. 25-26 ; EDA Kaggle G4 : la réussite baisse de Q1 à Q4 et chute dans la dernière minute).

Un quart-temps dure 12 min = 720 s, une prolongation 5 min. Ces bornes sont vérifiées dans les données avant le calcul.

```python
print("MINS_LEFT max par QUARTER :", d.groupby("QUARTER")["MINS_LEFT"].max().to_dict())
print("Répartition des QUARTER   :", d["QUARTER"].value_counts().sort_index().to_dict())

d["temps_restant_qt"] = d["MINS_LEFT"] * 60 + d["SECS_LEFT"]                 # secondes restantes dans le quart-temps
d["prolongation"]     = (d["QUARTER"] >= 5).astype(int)                     # 1 si prolongation (QUARTER 5 à 8)

# Temps restant dans le match : pour les 4 quart-temps réglementaires, on ajoute les quart-temps encore à jouer (720 s chacun).
# En prolongation, le "match" se résume à la prolongation en cours : on garde le temps restant du quart-temps.
d["temps_restant_match"] = np.where(d["QUARTER"] <= 4,
                                    (4 - d["QUARTER"]) * 720 + d["temps_restant_qt"],
                                    d["temps_restant_qt"])

d = d.drop(columns=["MINS_LEFT", "SECS_LEFT"])                               # remplacées par temps_restant_qt

derniere_minute = (d["QUARTER"] >= 4) & (d["temps_restant_qt"] <= 60)
print()
print(f"Réussite dans la dernière minute (Q4 ou prolongation) : {d.loc[derniere_minute, 'SHOT_MADE'].mean()*100:.1f} %  "
      f"vs reste du match : {d.loc[~derniere_minute, 'SHOT_MADE'].mean()*100:.1f} %")
```

**Lecture.** Les bornes sont cohérentes : 12 minutes au plus en quart-temps réglementaire, 4 minutes au plus (donc 5 min de jeu) en prolongation ; 0,6 % des tirs ont lieu en prolongation. L'écart « dernière minute » (42,0 % contre 46,6 %) retrouve le constat de l'EDA Kaggle sur un autre dataset. Le temps est désormais une variable continue.

### 4.2 Famille de geste

`ACTION_TYPE` compte 53 libellés dans la fenêtre, très déséquilibrés (« Jump Shot » 744 782 fois, « Running Hook Shot » 5 fois). Un libellé NBA est composé d'un **geste de base** et de **qualificatifs**. Exemple : *Driving Floating Bank Jump Shot* = geste « floater » (tir en suspension haute et courte, à une main), qualificatif « driving » (en pénétration), « bank » (avec la planche). Le geste de base devient `type_geste` :

| Famille `type_geste` | Mots-clés dans le libellé | Définition |
|---|---|---|
| Dunk | *dunk* | ballon claqué dans le cercle |
| Layup | *layup*, *finger roll* | tir déposé près du cercle (le finger roll est un layup « roulé » du bout des doigts) |
| Floater | *floating* | tir en cloche à courte distance, par-dessus un défenseur |
| Hook | *hook* | bras roulé, de côté |
| Jump shot | tout le reste | tir en suspension, y compris les variantes *bank*, *fadeaway*, *pullup*… |

Les **qualificatifs** (*driving*, *pullup*, *tip*, *alley oop*…) ne sont pas transformés en variables : la section 4.2 bis en donne la raison, mesures à l'appui.

```python
def type_geste(libelle):
    """Ramène un libellé ACTION_TYPE à l'une des 5 familles de geste."""
    a = libelle.lower()                          # insensible à la casse
    if "dunk" in a:                              return "Dunk"
    if "layup" in a or "finger roll" in a:       return "Layup"
    if "floating" in a:                          return "Floater"
    if "hook" in a:                              return "Hook"
    return "Jump shot"

d["type_geste"] = d["ACTION_TYPE"].map(type_geste)


# Bilan par famille : effectif, réussite, distance médiane
bilan = (d.groupby("type_geste")
          .agg(nb_tirs=("SHOT_MADE", "size"), reussite_pct=("SHOT_MADE", "mean"), distance_mediane_ft=("SHOT_DISTANCE", "median"))
          .sort_values("reussite_pct"))
bilan["reussite_pct"] = (bilan["reussite_pct"] * 100).round(1)
print(bilan.to_string())
```

**Lecture.** Le gradient est net et physiquement cohérent : Jump shot 37,8 % (distance médiane 24 ft) → Floater 45,2 % → Hook 48,2 % → Layup 56,0 % → Dunk 89,4 %. La famille de geste porte à la fois le geste *et* la distance, d'où son pouvoir discriminant ; `SHOT_DISTANCE` est conservée pour que le modèle puisse séparer les deux.

La cellule suivante affiche la table complète 53 → 5, pour vérification du regroupement.

```python
table = (d.groupby(["type_geste", "ACTION_TYPE"]).size().reset_index(name="nb_tirs")
          .sort_values(["type_geste", "nb_tirs"], ascending=[True, False]))
print(table.to_string(index=False))
```

### 4.2 bis Les qualificatifs du libellé : testés, puis écartés

Les qualificatifs d'`ACTION_TYPE` semblent porter une information sur la difficulté du tir : *driving* / *running* / *cutting* (tireur en mouvement vers le panier), *pullup* / *step back* / *fadeaway* / *turnaround* (tir créé seul, hors équilibre), *tip* / *putback* (tir immédiat après un rebond offensif), *alley oop* (passe lobée conclue en l'air). Quatre indicateurs 0/1 sont envisageables.

Avant de créer une variable, on vérifie qu'elle apporte une information **en plus** de ce que la table contient déjà. Un *Driving Layup Shot* réussit mieux qu'un *Jump Shot*, mais surtout parce qu'il est pris à deux pieds du panier, ce que `type_geste` et `SHOT_DISTANCE` disent déjà. L'écart de réussite avec / sans qualificatif est donc mesuré **à famille de geste et distance égales**, sur les saisons qui formeront le jeu d'entraînement (≤ 2022-23, découpage en 5.3), sans regarder le test.

```python
libelle = d["ACTION_TYPE"].str.lower()                                # version minuscule, calculée une fois
candidats = {                                                         # indicateurs CANDIDATS, calculés ici pour le test seulement
    "en_course":       libelle.str.contains("driving|running|cutting"),
    "en_desequilibre": libelle.str.contains("pullup|pull-up|step back|fadeaway|turnaround"),
    "seconde_chance":  libelle.str.contains("tip|putback"),
    "alley_oop":       libelle.str.contains("alley oop"),
}
futur_train = d["saison_debut"] <= 2022                               # saisons du futur jeu train (5.3)
tranches = pd.cut(d["SHOT_DISTANCE"], bins=[-1, 1, 4, 9, 15, 21, 24, 99],
                  labels=["0-1 ft", "2-4 ft", "5-9 ft", "10-15 ft", "16-21 ft", "22-24 ft", "25+ ft"])

lignes = []
for nom, ind in candidats.items():
    t = pd.DataFrame({"ind": ind[futur_train].astype(int), "geste": d.loc[futur_train, "type_geste"],
                      "tranche": tranches[futur_train], "made": d.loc[futur_train, "SHOT_MADE"]})
    brut = t.loc[t["ind"] == 1, "made"].mean() - t.loc[t["ind"] == 0, "made"].mean()          # écart sans aucun contrôle

    # Écart à geste ET distance égaux : dans chaque case (famille, tranche) ayant >= 1 000 tirs de chaque côté,
    # on compare la réussite avec / sans le qualificatif, puis on moyenne les cases (pondération de Mantel-Haenszel)
    cases = []
    for _, s in t.groupby(["geste", "tranche"], observed=True):
        n1, n0 = (s["ind"] == 1).sum(), (s["ind"] == 0).sum()
        if n1 >= 1000 and n0 >= 1000:
            cases.append((s.loc[s["ind"] == 1, "made"].mean() - s.loc[s["ind"] == 0, "made"].mean(), n1 * n0 / (n1 + n0)))
    ecarts, poids = zip(*cases)
    lignes.append((nom, int((t["ind"] == 1).sum()), round(brut * 100, 1), round(np.average(ecarts, weights=poids) * 100, 1), len(cases)))

print(pd.DataFrame(lignes, columns=["indicateur", "tirs_concernes", "ecart_brut_pts", "ecart_a_geste_et_distance_egaux_pts", "nb_cases"])
        .to_string(index=False))
del libelle, candidats, futur_train, tranches, lignes, t, cases
```

**Lecture.** Les écarts bruts comparent par exemple un *Pullup Jump shot* à **tous** les autres tirs, layups et dunks compris : ce sont des effets de composition. À famille de geste et distance égales, deux signes s'inversent : un tir « en déséquilibre » réussit **mieux** qu'un *Jump Shot* simple de même distance (+6 points), et un *tip* ou *putback* réussit **moins bien** qu'un layup ou un dunk posé (−6 points). L'information portée par ces qualificatifs existe, mais elle n'est pas celle que leur nom suggère, et elle est en grande partie déjà contenue dans `type_geste`, `SHOT_DISTANCE` et les coordonnées.

Ce test descriptif ne suffit pas à trancher : seule une mesure avec un modèle dit si ces colonnes améliorent la prédiction. Cette mesure a été faite (même modèle, avec et sans les quatre colonnes, entraîné sur `train`, évalué sur `validation`) : aucun gain, alors que retirer `type_geste` dégrade nettement le score.

**Décision : ces quatre variables ne sont pas créées.** `ACTION_TYPE` reste dans la table lisible comme colonne de contexte ; l'hypothèse pourra être re-testée à l'étape d'optimisation.

### 4.3 Tireur à domicile ou à l'extérieur

La table donne `HOME_TEAM` et `AWAY_TEAM` (abréviations, ex. `BOS`) mais l'équipe du tireur est en `TEAM_NAME` (ex. `Boston Celtics`) : rien n'indique directement si le tireur joue chez lui (EDA #7). Il faut une correspondance équipe → abréviation, absente des données.

Reconstruction sans table externe : dans un match, les tireurs appartiennent à deux équipes, l'une à domicile, l'autre à l'extérieur. Sur **tous les matchs d'une équipe dans une saison**, son abréviation apparaît dans `HOME_TEAM` ou `AWAY_TEAM` de **100 % de ses matchs**, alors que celles des adversaires n'apparaissent que quelques fois chacune. L'abréviation présente dans tous les matchs est donc la sienne. Le calcul est fait par saison, parce que des abréviations changent avec les déménagements (EDA var. 11).

```python
# Une ligne par (saison, équipe, match) avec les deux abréviations du match
matchs = (d.groupby(["saison_debut", "TEAM_ID", "GAME_ID"])[["HOME_TEAM", "AWAY_TEAM"]].first().reset_index())

# Empilement de HOME_TEAM et AWAY_TEAM dans une colonne abb : deux lignes par match de l'équipe
empile = pd.concat([matchs[["saison_debut", "TEAM_ID", "GAME_ID", "HOME_TEAM"]].rename(columns={"HOME_TEAM": "abb"}),
                    matchs[["saison_debut", "TEAM_ID", "GAME_ID", "AWAY_TEAM"]].rename(columns={"AWAY_TEAM": "abb"})])

# Combien de fois chaque abréviation apparaît dans les matchs de l'équipe, et combien de matchs elle a joués
compte  = empile.groupby(["saison_debut", "TEAM_ID", "abb"]).size().reset_index(name="n_apparitions")
n_match = matchs.groupby(["saison_debut", "TEAM_ID"]).size().reset_index(name="n_matchs")
compte  = compte.merge(n_match, on=["saison_debut", "TEAM_ID"])

# L'abréviation de l'équipe = celle présente dans TOUS ses matchs
mapping = compte[compte["n_apparitions"] == compte["n_matchs"]].set_index(["saison_debut", "TEAM_ID"])["abb"]

print("Équipes-saisons à mapper :", len(n_match), "| abréviations trouvées :", len(mapping),
      "| ambiguïtés (2 abréviations pour une équipe) :", mapping.index.duplicated().sum())

# Application : abréviation du tireur, puis indicateur domicile
d["abb_tireur"]      = pd.MultiIndex.from_frame(d[["saison_debut", "TEAM_ID"]]).map(mapping)
d["tireur_domicile"] = (d["abb_tireur"] == d["HOME_TEAM"]).astype(int)

# Contrôles : (a) chaque tireur est soit l'équipe à domicile soit à l'extérieur ; (b) chaque match a des tireurs des deux côtés
coherent = ((d["abb_tireur"] == d["HOME_TEAM"]) | (d["abb_tireur"] == d["AWAY_TEAM"])).mean()
deux_cotes = (d.groupby("GAME_ID")["tireur_domicile"].nunique() == 2).mean()
controle(f"tireur = équipe domicile ou extérieur dans {coherent*100:.1f} % des tirs", coherent == 1.0)
controle(f"les 2 équipes tirent dans {deux_cotes*100:.1f} % des matchs", deux_cotes == 1.0)

print()
print("Correspondance nom → abréviation (une par franchise, avec les renommages) :")
noms = d.groupby(["saison_debut", "TEAM_ID"])["TEAM_NAME"].first().to_frame().join(mapping)
print(noms.reset_index().groupby("TEAM_NAME")["abb"].unique().apply(lambda x: ", ".join(x)).to_string())
print()
print(d.groupby("tireur_domicile")["SHOT_MADE"].agg(nb_tirs="size", reussite="mean").round(4))
d = d.drop(columns=["abb_tireur"])                                            # variable intermédiaire, plus utile
```

**Lecture.** 300 équipes-saisons (30 équipes × 10 saisons), 300 abréviations trouvées, 0 ambiguïté, et les deux contrôles passent à 100 % : la correspondance est exacte. « LA Clippers » et « Los Angeles Clippers » sont la même franchise (`LAC`), ce qui confirme EDA var. 4 : `TEAM_NAME` n'est pas une clé fiable, `TEAM_ID` l'est.

L'avantage du terrain existe mais reste **modeste** : 46,9 % à domicile contre 46,0 % à l'extérieur, soit +0,8 point. Sur 2 millions de tirs l'écart est statistiquement significatif ; la variable restera d'appoint pour le modèle.

### 4.4 Coordonnées : contrôle de cohérence et correction

`LOC_X`, `LOC_Y` (pieds) et `SHOT_DISTANCE` (pieds, entier) décrivent la même chose : l'endroit d'où le tir est parti. L'EDA a mesuré une corrélation de 0,82 entre `SHOT_DISTANCE` et la distance recalculée depuis les coordonnées, attribuée à la rupture de collecte d'avant 2011. La fenêtre commençant en 2015, la corrélation devrait être quasi parfaite. Elle est recalculée, avec le panier en `(0 ; 5,25)` comme indiqué dans le template.

```python
d["dist_calc"] = np.sqrt(d["LOC_X"] ** 2 + (d["LOC_Y"] - 5.25) ** 2)        # Pythagore, panier à (0 ; 5,25)
ecart = (d["dist_calc"] - d["SHOT_DISTANCE"]).abs()

print(f"Corrélation dist_calc ↔ SHOT_DISTANCE : {d[['dist_calc', 'SHOT_DISTANCE']].corr().iloc[0, 1]:.4f}")
print(f"Écart absolu moyen                     : {ecart.mean():.2f} ft   |  tirs avec écart > 3 ft : {(ecart > 3).mean()*100:.1f} %")
print()
# Le même calcul saison par saison, sur les tirs à ≥ 10 ft pour que le ratio ait un sens
loin = d[d["SHOT_DISTANCE"] >= 10]
par_saison = loin.groupby("saison_debut").apply(
    lambda s: pd.Series({"corr":            np.corrcoef(s["dist_calc"], s["SHOT_DISTANCE"])[0, 1],
                         "ratio_median":    (s["SHOT_DISTANCE"] / s["dist_calc"]).median(),
                         "ecart_>3ft_pct":  ((s["dist_calc"] - s["SHOT_DISTANCE"]).abs() > 3).mean() * 100,
                         "LOC_X_abs_median": s["LOC_X"].abs().median()}),
    include_groups=False).round(3)
print(par_saison.to_string())
```

**Lecture.** Globalement, corrélation 0,736 et 3,7 ft d'écart moyen : *pire* que sur l'historique complet. Le tableau par saison localise le problème : sur 7 saisons la corrélation vaut 0,998-0,999 et le ratio `SHOT_DISTANCE / dist_calc` vaut 0,98 (l'entier est la partie entière du réel). Sur **trois saisons, 2019-20, 2020-21 et 2021-22, le ratio vaut 8,3 et 100 % des tirs sont concernés**. La `LOC_X` médiane y est de 1,3 ft au lieu de 13 : les coordonnées sont **environ 10 fois trop petites**.

C'est un défaut de **la source**, pas du fichier local : tout se passe comme si la conversion « dixièmes de pied → pieds, puis décalage du panier de 5,25 » avait été appliquée deux fois sur ces trois saisons. Le contrôle « coordonnées hors terrain = 0 » de l'EDA ne pouvait pas le détecter : des valeurs 10 fois trop petites restent dans le terrain.

**Correction.** Si une conversion `x → x/10` et `y → y/10 + 5,25` a été appliquée une fois de trop, l'inverse se déduit en imposant que les tirs au cercle retombent en `(0 ; ≈5,3)` comme dans les saisons saines : `LOC_X ← 10·LOC_X` et `LOC_Y ← 10·LOC_Y − 52,5`. La correction est appliquée **uniquement** aux trois saisons touchées, puis re-mesurée.

```python
saisons_touchees = [2019, 2020, 2021]
touche = d["saison_debut"].isin(saisons_touchees)
print("Avant correction — tir au cercle (SHOT_DISTANCE = 0), LOC_Y médiane :",
      "saison saine 2018 =", d.loc[(d['saison_debut'] == 2018) & (d['SHOT_DISTANCE'] == 0), 'LOC_Y'].median(),
      "| saison touchée 2019 =", d.loc[(d['saison_debut'] == 2019) & (d['SHOT_DISTANCE'] == 0), 'LOC_Y'].median())

d["LOC_X"] = np.where(touche, d["LOC_X"] * 10,        d["LOC_X"]).round(2)
d["LOC_Y"] = np.where(touche, d["LOC_Y"] * 10 - 52.5, d["LOC_Y"]).round(2)
d["dist_calc"] = np.sqrt(d["LOC_X"] ** 2 + (d["LOC_Y"] - 5.25) ** 2)
ecart = (d["dist_calc"] - d["SHOT_DISTANCE"]).abs()

print()
controle(f"corrélation après correction = {d[['dist_calc', 'SHOT_DISTANCE']].corr().iloc[0, 1]:.4f}", d[["dist_calc", "SHOT_DISTANCE"]].corr().iloc[0, 1] > 0.999)
controle(f"écart moyen = {ecart.mean():.2f} ft, écart > 3 ft sur {(ecart > 3).mean()*100:.2f} % des tirs", (ecart > 3).mean() < 0.001)
hors = ((d["LOC_X"].abs() > 25) | (d["LOC_Y"] < 0) | (d["LOC_Y"] > 94)).sum()
controle(f"coordonnées hors terrain = {hors}", hors == 0)
controle(f"SHOT_DISTANCE = partie entière de dist_calc sur {(np.floor(d['dist_calc']) == d['SHOT_DISTANCE']).mean()*100:.1f} % des tirs",
         (np.floor(d["dist_calc"]) == d["SHOT_DISTANCE"]).mean() > 0.99)

print()
print("Par saison après correction :")
print(d[d["SHOT_DISTANCE"] >= 10].groupby("saison_debut").apply(
        lambda s: pd.Series({"corr": np.corrcoef(s["dist_calc"], s["SHOT_DISTANCE"])[0, 1],
                             "LOC_X_abs_median": s["LOC_X"].abs().median()}), include_groups=False).round(3).to_string())

d = d.drop(columns=["dist_calc"])        # a servi au contrôle ; SHOT_DISTANCE (officiel) + LOC_X/LOC_Y (fins) suffisent
```

**Lecture.** Après correction : corrélation 0,9996, écart moyen 0,48 ft, aucun tir hors terrain, et `SHOT_DISTANCE` est la partie entière de la distance recalculée dans 99,8 % des cas, sur les dix saisons. Les coordonnées sont désormais homogènes ; un *shot chart* multi-saisons devient possible.

Conséquences :
- toute analyse spatiale faite sur 2019-20 à 2021-22 **sans cette correction** est fausse (cartes de tir, densités) ; les analyses par distance ou par zone ne sont pas touchées (`SHOT_DISTANCE` et `BASIC_ZONE` sont corrects) ;
- à documenter dans le rapport comme difficulté rencontrée, et éventuellement à signaler au mainteneur du dépôt source.

`SHOT_DISTANCE` (entier officiel) et `LOC_X` / `LOC_Y` (position fine, qui distingue un corner d'un tir dans l'axe à même distance) sont conservés ; `dist_calc`, redondante, est retirée.

### 4.5 Poste du joueur : imputer, pas supprimer

`POSITION_GROUP` (G / F / C) manque sur toute la saison 2024-25 (le fichier source n'a pas la colonne) et sur quelques centaines de tirs par saison depuis 2017 (EDA #2). Supprimer ces lignes retirerait la saison la plus récente, celle que le modèle doit prédire.

Le poste d'un joueur est **quasi stable** d'une saison à l'autre. Pour un tir sans poste, on prend donc le **dernier poste connu du même joueur** (trié par date). Les joueurs sans aucun poste renseigné, typiquement les *rookies* de 2024-25, reçoivent la modalité explicite `Inconnu` : c'est une information en soi (« première saison »), exploitable par le modèle.

```python
print("Postes manquants par saison :", d.groupby("saison_debut")["POSITION_GROUP"].apply(lambda s: s.isna().sum()).to_dict())
print("Répartition :", d["POSITION_GROUP"].value_counts(dropna=False).to_dict())

# Dernier poste connu par joueur : on ne garde que les lignes renseignées, on trie par date, on prend la dernière
dernier_poste = (d.dropna(subset=["POSITION_GROUP"])
                   .sort_values("GAME_DATE")
                   .groupby("PLAYER_ID")["POSITION_GROUP"].last())

# Stabilité du poste : combien de joueurs ont eu plus d'un POSITION_GROUP dans la fenêtre ?
nb_postes = d.dropna(subset=["POSITION_GROUP"]).groupby("PLAYER_ID")["POSITION_GROUP"].nunique()
print(f"Joueurs ayant changé de POSITION_GROUP sur la période : {(nb_postes > 1).sum()} / {len(nb_postes)} ({(nb_postes > 1).mean()*100:.0f} %)")

manquant = d["POSITION_GROUP"].isna()
n_manquant = manquant.sum()
d.loc[manquant, "POSITION_GROUP"] = d.loc[manquant, "PLAYER_ID"].map(dernier_poste)     # imputation
reste = d["POSITION_GROUP"].isna()
print(f"Manquants : {n_manquant:,} → imputés par le dernier poste connu : {n_manquant - reste.sum():,} → restants : {reste.sum():,} "
      f"({d.loc[reste, 'PLAYER_ID'].nunique()} joueurs sans aucun poste connu)".replace(",", " "))
print("Exemples de joueurs sans poste connu :", d.loc[reste, "PLAYER_NAME"].value_counts().head(6).index.tolist())

d["POSITION_GROUP"] = d["POSITION_GROUP"].fillna("Inconnu")
print()
print(d.groupby("POSITION_GROUP")["SHOT_MADE"].agg(nb_tirs="size", reussite="mean").round(3))
```

**Lecture.** 226 044 postes manquants dans la fenêtre, dont 219 000 pour 2024-25. L'imputation par le dernier poste connu en résout 204 208 ; les 21 836 restants (117 joueurs) sont des rookies 2024-25 (Risacher, Castle, Sarr, Carrington…) sans saison antérieure. Ils deviennent `Inconnu`.

Le poste est un signal : les pivots (C) réussissent 54,3 % de leurs tirs (tirs près du cercle), les arrières (G) 43,9 % (tirs lointains). 14 % des joueurs changent de groupe sur dix ans : l'imputation est une approximation raisonnable, pas une vérité.

## 5. Clé du tir, table propre, découpage, encodage, sauvegarde

### 5.1 Une clé pour chaque tir

Le dataset Kaggle avait un identifiant d'événement (`Game Event ID`) ; celui-ci n'en a pas. Une clé est nécessaire pour trois usages : retrouver un tir précis lors de l'examen d'une prédiction, joindre d'autres tables (play-by-play, box score), et **relier les deux fichiers de sortie** (table lisible et table encodée).

Deux options. La **clé naturelle**, combinaison de colonnes existantes censée identifier un tir, ici (`GAME_ID`, `QUARTER`, `temps_restant_qt`, `PLAYER_ID`) : à vérifier, car un joueur peut tirer deux fois dans la même seconde (tir raté, rebond, re-tir immédiat). La **clé technique**, un numéro de ligne `shot_id` attribué après le nettoyage : sans signification, mais garanti unique. La première est mesurée, la seconde adoptée.

```python
cle_naturelle = ["GAME_ID", "QUARTER", "temps_restant_qt", "PLAYER_ID"]
n_doublons_cle = d.duplicated(subset=cle_naturelle).sum()                  # tirs partageant la même clé naturelle qu'un autre
print(f"Clé naturelle {cle_naturelle} : {n_doublons_cle:,} collisions sur {len(d):,} tirs".replace(",", " "))

# À quoi ressemblent ces collisions ? (même joueur, même seconde)
ex = d[d.duplicated(subset=cle_naturelle, keep=False)].sort_values(cle_naturelle)
print(ex[["GAME_ID", "QUARTER", "temps_restant_qt", "PLAYER_NAME", "ACTION_TYPE", "SHOT_MADE", "SHOT_DISTANCE"]].head(6).to_string(index=False))

# Clé technique : numéro de ligne, de 1 à n, après tri chronologique (saison, match, quart-temps, temps décroissant)
d = d.sort_values(["saison_debut", "GAME_ID", "QUARTER", "temps_restant_qt"], ascending=[True, True, True, False]).reset_index(drop=True)
d.insert(0, "shot_id", np.arange(1, len(d) + 1))                            # en première colonne
controle(f"shot_id unique ({d['shot_id'].nunique():,} valeurs pour {len(d):,} lignes)".replace(",", " "), d["shot_id"].is_unique)
```

### 5.2 Trois familles de colonnes

Chaque colonne est rangée dans l'une de trois familles. Cette classification garantit la neutralité du modèle décidée au cadrage : les identifiants (joueur, équipe, match) restent dans la table pour l'analyse *après* modélisation (réel − attendu par joueur), mais **ne sont pas** des variables d'entrée.

| Famille | Colonnes | Rôle |
|---|---|---|
| Identifiants & contexte | `shot_id`, `saison_debut`, `SEASON_1`, `GAME_ID`, `GAME_DATE`, `TEAM_ID`, `TEAM_NAME`, `PLAYER_ID`, `PLAYER_NAME`, `HOME_TEAM`, `AWAY_TEAM`, `ACTION_TYPE`, `SHOT_TYPE`, `ZONE_NAME`, `ZONE_RANGE` | retrouver, filtrer, agréger ; jamais en entrée du modèle |
| Variables explicatives numériques | `SHOT_DISTANCE`, `LOC_X`, `LOC_Y`, `tir_3pts`, `QUARTER`, `prolongation`, `temps_restant_qt`, `temps_restant_match`, `tireur_domicile` | entrée du modèle |
| Variables explicatives catégorielles | `BASIC_ZONE`, `type_geste`, `POSITION_GROUP` | entrée du modèle, après encodage |
| Cible | `SHOT_MADE` | variable à prédire |

`saison_debut` est classée en identifiant ; le cadrage prévoit de la tester comme variable (« effet époque », EDA #3) à l'étape d'optimisation. Elle reste donc accessible dans les deux tables.

```python
identifiants = ["shot_id", "saison_debut", "SEASON_1", "GAME_ID", "GAME_DATE", "TEAM_ID", "TEAM_NAME", "PLAYER_ID", "PLAYER_NAME",
                "HOME_TEAM", "AWAY_TEAM", "ACTION_TYPE", "SHOT_TYPE", "ZONE_NAME", "ZONE_RANGE"]
features_num = ["SHOT_DISTANCE", "LOC_X", "LOC_Y", "tir_3pts", "QUARTER", "prolongation", "temps_restant_qt",
                "temps_restant_match", "tireur_domicile"]
features_cat = ["BASIC_ZONE", "type_geste", "POSITION_GROUP"]
cible        = ["SHOT_MADE"]

propre = d[identifiants + features_num + features_cat + cible].copy()
print("Table propre :", f"{propre.shape[0]:,} lignes × {propre.shape[1]} colonnes".replace(",", " "))
print()
print(propre.isna().sum().sum(), "valeur(s) manquante(s) au total")
print()
print(propre.head(3).T)
```

### 5.3 Découpage train / validation / test **par saison**

Un découpage aléatoire mélangerait des tirs de 2024 dans l'entraînement et de 2016 dans le test : le modèle serait évalué sur un passé qu'il a déjà vu, et son score serait trop optimiste. En production, le modèle prédit des saisons qu'il n'a pas vues : le découpage se fait donc dans le temps (EDA #3, cadrage).

| Jeu | Saisons | Usage |
|---|---|---|
| `train` | 2015-16 → 2022-23 (8 saisons) | apprendre |
| `validation` | 2023-24 | comparer les modèles et régler leurs paramètres |
| `test` | 2024-25 | mesurer **une seule fois**, à la fin ; c'est aussi la saison où les postes sont imputés, donc une mesure en conditions réelles |

Une colonne `split` est ajoutée plutôt que trois fichiers créés : une seule table, un seul filtre à appliquer.

```python
propre["split"] = np.select([propre["saison_debut"] <= 2022, propre["saison_debut"] == 2023],
                            ["train", "validation"],
                            default="test")

resume = (propre.groupby("split")
            .agg(nb_tirs=("SHOT_MADE", "size"),
                 saisons=("saison_debut", lambda s: f"{s.min()}-{s.max()}"),
                 reussite_pct=("SHOT_MADE", "mean"),
                 part_3pts_pct=("tir_3pts", "mean"))
            .loc[["train", "validation", "test"]])
resume[["reussite_pct", "part_3pts_pct"]] = (resume[["reussite_pct", "part_3pts_pct"]] * 100).round(1)
print(resume.to_string())
```

**Lecture.** 1 658 384 tirs pour apprendre, 218 254 pour valider, 219 000 pour tester. Le taux de réussite est stable (46,3 → 47,5 → 46,8 %) : le plancher « prédire la moyenne de `train` » donnera environ 46 % de réussite attendue pour tout tir, et tout modèle devra faire mieux. La part de 3 points continue de monter (35,6 → 39,4 → 42,0 %) : le jeu change à l'intérieur même de la fenêtre, ce qui motive le test de `saison_debut` comme variable.

### 5.4 Encodage des variables catégorielles

Un modèle ne lit que des nombres. `BASIC_ZONE` vaut « Mid-Range » ou « Restricted Area » : une traduction est nécessaire. Deux méthodes classiques :

- **encodage ordinal** (Mid-Range = 1, Restricted Area = 2…) : inadapté ici, il impose un ordre qui n'existe pas (« Restricted Area » n'est pas « deux fois » Mid-Range) ;
- **one-hot** : une colonne 0/1 par modalité (`BASIC_ZONE_Mid-Range`, `BASIC_ZONE_Restricted Area`…). Pas d'ordre artificiel, lisible. C'est la méthode retenue, avec `pd.get_dummies`.

Coût : 3 colonnes catégorielles (6 + 5 + 4 modalités) deviennent 15 colonnes 0/1. Avec les 53 `ACTION_TYPE` bruts il y en aurait eu 53 : c'est l'intérêt du regroupement de la section 4.2.

Les variables déjà numériques et les indicateurs 0/1 ne sont pas modifiés. Aucune mise à l'échelle (standardisation) n'est appliquée ici : elle dépend du modèle choisi (nécessaire pour une régression logistique, inutile pour un arbre) et doit être ajustée sur `train` seulement ; elle est faite dans le notebook de modélisation, dans un `Pipeline`.

```python
# One-hot (une colonne 0/1 par modalité) : aucun ordre n'existe entre les modalités de BASIC_ZONE, type_geste, POSITION_GROUP.
# get_dummies sur la table complète : les 15 modalités sont présentes dans train, validation et test (mêmes colonnes partout).
X = pd.get_dummies(propre[features_num + features_cat], columns=features_cat, dtype=int)   # dtype=int → 0/1 et non True/False

encode = pd.concat([propre[["shot_id", "saison_debut", "GAME_ID", "PLAYER_ID", "split", "SHOT_MADE"]],   # clés + cible pour retrouver / filtrer
                    X], axis=1)

print("Variables d'entrée du modèle :", X.shape[1])
print(list(X.columns))
print()
print(encode.head(3).T)
```

### 5.5 Sauvegarde

Deux fichiers dans `data/processed/` (ignoré par Git : chacun les régénère en exécutant ce notebook) :

- `shots_clean_2015_2025.parquet` : la table **lisible** (libellés en clair), pour les analyses descriptives et la comparaison réel − attendu par joueur ;
- `shots_encoded_2015_2025.parquet` : la table **prête pour le modèle** (clés + cible + variables 0/1 et numériques).

Format **Parquet** plutôt que CSV : 5 à 10 fois plus compact, types conservés (une date reste une date), lecture quasi instantanée. Il requiert la bibliothèque `pyarrow`, présente dans `requirements.txt`. Les fichiers existants sont remplacés à chaque exécution.

```python
def sauvegarder(table, nom):
    """Écrit en Parquet si pyarrow est disponible, sinon en CSV. Retourne le chemin écrit."""
    try:
        chemin = OUT / f"{nom}.parquet"
        table.to_parquet(chemin, index=False)
    except ImportError:                                      # repli en CSV si pyarrow n'est pas installé
        chemin = OUT / f"{nom}.csv"
        table.to_csv(chemin, index=False)
        print("pyarrow absent : écriture en CSV.")
    print(f"{chemin}  —  {chemin.stat().st_size / 1e6:.1f} Mo  —  {table.shape[0]:,} lignes × {table.shape[1]} colonnes".replace(",", " "))
    return chemin

chemin_clean  = sauvegarder(propre, "shots_clean_2015_2025")
chemin_encode = sauvegarder(encode, "shots_encoded_2015_2025")
```
